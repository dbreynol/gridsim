```python
import numpy as np
import pandas as pd
from scipy.optimize import linprog
```

```python
K = np.array( [[1,-1,0], [1,0,1], [0,1,-1] ])
thetas_f = np.array( [ [2/3, 1/3], [1/3, 2/3], [0,0] ] )

ptdf = K @ thetas_f

capacities = np.array([10, 10, 100]) # line capacities
costs = np.array([10, 50])
A_ub = ptdf
b_ub = capacities

A_eq = [[1, 1]]
b_eq = [16]
#bounds = [(0, 10), (0, 8), (0,12), (0,16)]

result = linprog(
    c=costs,
    A_ub=A_ub,
    b_ub=b_ub,
    A_eq=A_eq,
    b_eq=b_eq,
    #bounds=bounds,
    method="highs"
)

print(result.x)
print(result.fun)
```

```python
print(result.ineqlin.residual)
print(-result.ineqlin.marginals)
print(result.eqlin.marginals)
```
