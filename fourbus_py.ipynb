```python
import pandas as pd
import pypsa
```

```python
n = pypsa.Network()

# 1. Add all buses in one line
n.add("Bus", ["A", "B", "C"])

# 2. Add all lines at once (x=1 and r=1 automatically broadcast to all 3 lines)
n.add(
    "Line",
    ["A-B", "A-C", "B-C"],
    bus0=["A", "A", "B"],
    bus1=["B", "C", "C"],
    s_nom=[500, 10, 100],
    x=1,
    r=1,
)

# 3. Add both generators at once (p_nom=100 broadcasts to both)
n.add(
    "Generator",
    ["A", "B"],
    bus=["A", "B"],
    p_nom=100,
    marginal_cost=[10, 50],
)

# 4. Add load
n.add("Load", "Load at C", bus="C", p_set=20)

n.optimize(solver_name="highs", assign_all_duals=True)
```
